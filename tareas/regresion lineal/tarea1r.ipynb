3. (1 punto) **Construcción del modelo** Considere inicialmente las variables `lcavol, lweight, age, svi`.


*   Ajuste un modelo lineal para explicar lpsa a partir de estas cuatro variables. Utilice `df_train`.
*   Escriba la ecuación del modelo estimado usando **los coeficientes que obtuvo en R.**
* De las cuatro variables, indique cuál presenta mayor evidencia de asociación con lpsa.

Incluya código utilizado y el resultado numérico que justifíca su respuesta.

```r
# construcción del modelo lineal
modelo_lpsa <- lm(lpsa ~ lcavol + lweight + age + svi, data = df_train)
modelo_lpsa
```

```r
summary(modelo_lpsa)
```

**ecuación del modelo estimado con coeficientes obtenidos**

lpsa = -0.64219 + 0.76812\*lweight + 0.53173\*lcavol + 0.52891\*svi - 0.00784\*age

La variable con mayor evidencia de asociación con lpsa es "lcavol" (log-volumen del cáncer): su p-value es 8.13e-07, el más pequeño de las cuatro variables (t = 5.48).

5. (2 puntos) **¿Conviene complicar el modelo?** Ajuste ahora:

```r
modelo_2 <- lm(lpsa ~ lcavol + lweight + age + svi +lbph + lcp + gleason + pgg45,
               data = df_train)
```

Compare este modelo con el modelo anterior usadno:
* R^2 ajustado
* AIC
Si tuviera que escoger solo uno de los dos modelos ¿Cuál escogería?

```r
summary(modelo_lpsa)$adj.r.squared # para ver los r^2 ajustados
summary(modelo_2)$adj.r.squared
AIC(modelo_lpsa, modelo_2)
```

considerando los coeficientes obtenidos, el modelo 2 supera al modelo 1 (modelo_lpsa) al tener un mayor r^2 ajustado y un menor AIC, por lo que es el modelo elegido.

usamos el r^2 ajustado y no el normal ya que el primero permite evidenciar la penalización de agregar nuevas variables que empeoran el modelo.

Considere el siguiente código, que obtiene el error predictivo asociado al segundo modelo.

```r
pred_2 <- predict(modelo_2, newdata = df_test)
mean((df_test$lpsa - pred_2)^2)
```
